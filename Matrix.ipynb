# Adjoint of a Matrix

```python
M=np.array([  [1, 1, 1],
               [-2,-1, 1],
                [3, 1,-5]   ])

a=((-1)**(0+0)) * ( (-5 * -1) - (1*1)  )
b=((-1)**(0+1)) * ( (-2 * -5) - (1*3)  )
c=((-1)**(0+2)) * ( (-2 *  1) - (-1*3) )
d=((-1)**(1+0)) * ( ( 1 * -5) - (1*1)  )
e=((-1)**(1+1)) * ( ( 1 * -5) - (1*3)  )
f=((-1)**(1+2)) * ( ( 1 *  1) - (1*3)  )
g=((-1)**(2+0)) * ( ( 1 *  1) - (-1*1) )
h=((-1)**(2+1)) * ( ( 1 *  1) - (1*-2) )
i=((-1)**(2+2)) * ( ( 1 * -1) - (1*-2) )

adj=np.empty((3,3),  dtype=int)

adj[0][0]=a
adj[0][1]=b
adj[0][2]=c
adj[1][0]=d
adj[1][1]=e
adj[1][2]=f
adj[2][0]=g
adj[2][1]=h
adj[2][2]=i

print("Found_co-factors of the matrix M  \n\n", adj)
```

```python
print("Transposed the Co_factors to shape adjoint of matrix M \n\n", adj.T)
```

# Inverse of a Matrix using Adjoint Method

```python
I=np.array([  [1,0,2],
              [0,2,1],
             [1,-1,1]  ])    

# Inverse in adjoint method = adj/|I|
# where the determinant of the matrix should not be equal to zero

dt=1*( (2*1)-(-1*1) ) - 0*( (1*0)-(1*1) ) + 2*( (0*-1)-(1*2) )

dt
```

```python
j=((-1)**(0+0)) * ( (2*1) - (1*-1)  )
k=((-1)**(0+1)) * ( (0*1) - (1*1 )  )
l=((-1)**(0+2)) * ( (0*-1) - (2*1) )
m=((-1)**(1+0)) * ( (0*1) - (2*-1)  )
n=((-1)**(1+1)) * ( (1*1) - (2*1)  )
o=((-1)**(1+2)) * ( (1*-1) - (0*1)  )
p=((-1)**(2+0)) * ( (0*1) - (2*2) )
q=((-1)**(2+1)) * ( (1*1) - (0*2) )
r=((-1)**(2+2)) * ( (1*2) - (0*0) )

cf=np.empty((3,3),   dtype=float)

cf[0][0]=j
cf[0][1]=k
cf[0][2]=l
cf[1][0]=m
cf[1][1]=n
cf[1][2]=o
cf[2][0]=p
cf[2][1]=q
cf[2][2]=r

adj=cf.T
print("\nFound_co-factors of the matrix I  \n\n", cf)
print("\nTransposed the Co_factors to shape adjoint of matrix I \n\n",adj)
```

```python
print("\nFinal inverse answer of the matrix I  \n\n", adj/dt)
```

# System Of Equations¶
# Gauss elimination (forming Row Echelon Form)
# EQUATION
# 1x+1y+1z=6.
# 2x+3y+1z=10.
# 1x+2y+3z=13






# AX=B
# [1,1, 1] [x] [6]
# [2,3,1] [y] [10]
# [1,2,3] [z] [13]

```python
A=np.array([    [1,1,1],
                [2,3,1],
                [1,2,3] ])
B=np.array([  [6],
             [10],
             [13]   ])
aug=np.concatenate((A,B),   axis=1)
aug
```

```python
# >>  How to access the whole row and the whole column? 

#     first row = Matri[0]                 = R1.
#     2nd row = Matrix[1]                   = R2.
#     third row = Matrix[2]                 = R3.
#     first column = Matrix[ : , 0]         = C1.
#     2nd column = Matrix[ : , 1]           = C2.
#     third column = Matrix[ : , 2 ]        = C3.
```

```python
aug[1]=aug[1]-2*aug[0]                         # >>  R2 = R2-2R1
aug
```

```python
aug[2]=aug[2]-1*aug[0]                         # >>  R3 = R3-R1
aug
```

```python
aug[2]=aug[2]-1*aug[1]                         # >>  R3 = R3-R2
aug
```

```python
aug[2]=aug[2]/3                                # >>  R3 = R3/3.
print("\nProduced row echelon form !\n")
aug
```

# x=2 , y=1 , z=3 (Row echelon form values)
# Gauss-Jordan (forming Reduced Row Echelon Form), continuing the previous (row echelon) manipulation.

```python
aug[0]=aug[0]-1*aug[2]                         # >>  R1 = R1-1R3
aug
```

```python
aug[1]=aug[1]+1*aug[2]                         # >>  R2 = R2+1R3
aug
```

```python
aug[0]=aug[0]-1*aug[1]                         # >>  R1 = R1-1R2
aug
```
